# 1. Introducción

## 1.1 Contexto del Dataset

Este dataset proviene del Sistema de Vigilancia de Salud Publica (SIVIGILA), que recopila y analiza datos sobre eventos que afectan la salud de la poblacion colombiana: enfermedades infecciosas, cronicas, factores de riesgo y otros sucesos de importancia en salud publica.

El dengue es un evento de notificacion obligatoria en Colombia desde la reemergencia de los años setenta. La notificacion se realiza de forma individual por las Unidades Primarias Generadoras de Datos (UPGD) al sistema SIVIGILA. Se manejan 3 eventos de notificacion relacionados al dengue: dengue (210), dengue grave (220) y mortalidad por dengue (580).

El período de análisis comprende los años disponibles en fuentes abiertas del INS (2007-2024)

## 1.2 Descripción del Conjunto de Datos

Los datos se obtienen del portal oficial del INS - SIVIGILA:
https://portalsivigila.ins.gov.co/Paginas/Buscador.aspx

Cada fila representa un caso individual notificado con informacion demografica, geografica, clinica y administrativa.

> Diccionario oficial: https://www.ins.gov.co/BibliotecaDigital/diccionario-xls-.pdf

## 3.1 Perfil de Variables

Para cada variable se reporta el tipo de dato, el numero y porcentaje de valores nulos, y el numero de valores unicos. Esta información, junto con el diccionario de datos, permite identificar variables con problemas de calidad y fundamentar las decisiones de limpieza de la sección 4.

```python
# trata strings vacios o solo espacios como nulos
df_check = df_raw.apply(lambda col: col.map(lambda x: pd.NA if isinstance(x, str) and x.strip() == "" else x))

nulos     = df_check.isnull().sum()
pct_nulos = (df_check.isnull().mean() * 100).round(1)
unicos    = df_raw.nunique()

inventario = pd.DataFrame({
    "tipo":      df_raw.dtypes,
    "nulos":     nulos,
    "pct_nulos": pct_nulos,
    "unicos":    unicos,
})

inventario = inventario.sort_values("pct_nulos", ascending=False)
display(inventario)
```

El dataset crudo contiene 1.635.839 filas y 79 columnas, todas en formato texto. Ninguna tiene tipo numérico o fecha asignado por defecto, por lo que EDAD, SEMANA, ANO, estrato, todas las fechas y otras variables categóricas que se manejan con números requieren conversión explícita en la etapa de limpieza.

Las variables núcleo FEC_NOT, ANO, SEMANA, EDAD, SEXO, AREA, COD_EVE, INI_SIN, FEC_CON y todos los códigos geográficos están completas o casi completas, lo que garantiza la viabilidad del análisis principal.

Sin embargo se dectecta que hay otras variables de caracterización complementaria que presentan pocos datos por lo que ahora veremos el porcentaje de nulos a lo largo de los años para identificar si se trata de un problema estructural de los datos que este justificado

```python
# % de nulos por año, para detectar columnas que solo se dejaron de llenar en años puntuales
pct_nulos_por_ano = (
    df_check
    .groupby(df_raw["ANO"])
    .apply(lambda g: g.isnull().mean() * 100)
    .round(1)
    .T
)
pct_nulos_por_ano = pct_nulos_por_ano.loc[inventario.index]
display(pct_nulos_por_ano)
```

Los valores faltantes en FEC_HOS (61%), CBMTE y FEC_DEF (99%) son estructurales y justificados por subpoblación: cada variable aplica únicamente a un subconjunto específico de pacientes, hospitalizados para FEC_HOS y casos con COD_EVE=580 para las variables de mortalidad. Esta condición es independiente del año del registro y se mantiene constante en todo el rango 2007-2024.

Los valores faltantes en estrato, fuente, sem_ges y nom_grupo responden a cobertura temporal del instrumento: estrato, fuente y sem_ges se incorporaron a la ficha en 2019, y nom_grupo en 2018 reemplazando al campo codificado GRU_POB, que se diligenció entre 2007 y 2012 con un vacío real de captura entre 2013 y 2017. A esto se suma una restricción adicional en sem_ges, que solo aplica a mujeres en gestación, y en nom_grupo, que solo aplica a pacientes pertenecientes a una comunidad étnica específica. Estrato y fuente, en cambio, aplican a todo paciente y no tienen esta segunda restricción. Estas consideraciones deben tenerse en cuanta al momento de mostrar el análisis de caracterización para no mal interpretar los resultados.

En la etapa de limpieza se realizarán las siguientes acciones. Primero, conversión de todas las variables a su tipo de dato correcto. Segundo, descarte de variables sin valor analítico, ya sea por estar fuera del foco de estudio o por calidad de dato deficiente: CONSECUTIVE, CONSECUTIVE2, consecutive_12, Particion, Partición, Separación ,COD_EVE.1 ,FM_FUERZA, FM_UNIDAD, FM_GRADO, consecutive_origen, COD_SUB, COD_PRE, FEC_AJU, FEC_ARC_XL, va_sispro y CER_DEF. Tercero, descarte de variables que violan la regla de no transitividad, es decir, columnas de texto que duplican información ya contenida en sus códigos correspondientes: Pais_ocurrencia, Pais_residencia, Nombre_evento, Departamento_ocurrencia, Municipio_ocurrencia, Departamento_residencia, Municipio_residencia, Departamento_Notificacion, Municipio_notificacion, nom_est_f_caso, nombre_nacionalidad y FECHA_NTO. Cuarto, creación de la variable derivada edad_anios a partir de EDAD y UNI_MED, tras lo cual ambas variables originales serán descartadas, Quinto dado que vemos que la variable nom_grupo y GRU_POB se refieren a lo mismo debemos hacer el mapeo del codigo con el nombre del grupo etnico para poder representar todos los valores obtenidos

## 3.3 Distribución de COD_EVE

El campo COD_EVE define el tipo de evento notificado y determina que análisis aplica a cada registro: incidencia (210, 220) o mortalidad (580). Esta sección examina la proporcion de cada evento.

```python
conteo_cod_eve = (
    df_raw["COD_EVE"].astype(str).str.strip()
    .value_counts().reset_index()
)
conteo_cod_eve.columns = ["COD_EVE", "conteo"]
conteo_cod_eve["pct"] = (conteo_cod_eve["conteo"] / len(df_raw) * 100).round(2)

etiquetas = {"210": "Dengue (210)", "220": "Dengue grave (220)", "580": "Mortalidad (580)"}
conteo_cod_eve["label"] = conteo_cod_eve["COD_EVE"].map(etiquetas)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

wedges, _, autotexts = axes[0].pie(
    conteo_cod_eve["conteo"],
    labels=None,
    autopct="%1.1f%%",
    colors=[INS_AZUL, INS_VERDE, INS_ROJO],
    startangle=90,
    pctdistance=0.75,
    wedgeprops={"edgecolor": "white", "linewidth": 1.5}
)
for at in autotexts:
    at.set_color("white")
    at.set_fontsize(9)
axes[0].legend(
    wedges, conteo_cod_eve["label"],
    loc="lower center", bbox_to_anchor=(0.5, -0.12),
    fontsize=9, frameon=False
)
axes[0].set_title("Distribución por tipo de evento", fontsize=11)

detalle = conteo_cod_eve[conteo_cod_eve["COD_EVE"].isin(["220", "580"])]
axes[1].barh(detalle["label"], detalle["conteo"], color=[INS_VERDE, INS_ROJO], edgecolor="white")
for i, (val, pct) in enumerate(zip(detalle["conteo"], detalle["pct"])):
    axes[1].text(val + 200, i, f"{val:,} ({pct}%)", va="center", fontsize=9)
axes[1].set_xlabel("Registros")
axes[1].set_title("Detalle: dengue grave y mortalidad", fontsize=11)

plt.suptitle("Distibución de COD_EVE (210, 220, 580)", fontsize=12)
plt.tight_layout()
plt.savefig(DIR_FIGURAS / "distribucion_cod_eve.png", dpi=150, bbox_inches="tight")
plt.show()
```

## 3.4 Clasificación y Estado Final del Caso

Los campos TIP_CAS, AJUSTE, confirmados y Estado_final_de_caso determinan conjuntamente el ciclo de vida de un caso en SIVIGILA, desde su clasificación inicial hasta su estado definitivo tras el proceso de ajuste. El protocolo INS indica que deben excluirse los registros con AJUSTE = 6 (descartado) y AJUSTE = D (error de digitación) antes de cualquier análisis. Esta subsección documenta la distribución real de los tres campos y detecta valores inesperados o inconsistencias entre ellos.

## 3.6 Cobertura de Notificación por Municipio

Numero de semanas epidemiologicas con al menos un caso notificado por municipio y año. Un valor de 0 indica que el municipio no reporto ningun caso durante ese ano completo.

```python
_df = df_raw.copy()
_df["ANO"] = pd.to_numeric(_df["ANO"], errors="coerce")
_df["SEMANA"] = pd.to_numeric(_df["SEMANA"], errors="coerce")
_df = (
    _df[_df["COD_DPTO_O"].astype(str).str.lstrip("0") == "47"]
    .dropna(subset=["COD_MUN_O", "ANO", "SEMANA"])
)

anos_disponibles = sorted(_df["ANO"].unique().astype(int))

semanas_con_casos = (
    _df.groupby(["COD_MUN_O", "ANO", "SEMANA"])
    .size()
    .reset_index(name="casos")
    .groupby(["COD_MUN_O", "ANO"])
    .size()
    .reset_index(name="sem_con_casos")
)

pivot_cobertura = semanas_con_casos.pivot(
    index="COD_MUN_O", columns="ANO", values="sem_con_casos"
).fillna(0)

_muns = (
    pd.read_excel(DIR_MAPPINGS / "Davipola.xlsx", dtype=str)
    .assign(cod_municipio=lambda d: "47" + d["cod_municipio"].astype(str).str.strip().str[-3:])
    .query("cod_departamento.str.strip().str.lstrip('0') == '47'", engine="python")
    [["cod_municipio", "municipio"]]
    .drop_duplicates()
)

pivot_cobertura = pivot_cobertura.reset_index()
pivot_cobertura["COD_MUN_O"] = "47" + pivot_cobertura["COD_MUN_O"].astype(str).str.strip().str.zfill(3)
pivot_cobertura = pivot_cobertura.merge(_muns, left_on="COD_MUN_O", right_on="cod_municipio", how="left")
pivot_cobertura["municipio"] = pivot_cobertura["municipio"].fillna(pivot_cobertura["COD_MUN_O"])
pivot_cobertura = (
    pivot_cobertura
    .set_index("municipio")
    .drop(columns=["COD_MUN_O", "cod_municipio"], errors="ignore")
    .sort_values(anos_disponibles[-1], ascending=False)
)

fig, ax = plt.subplots(figsize=(len(anos_disponibles) + 3, len(pivot_cobertura) * 0.38 + 2))
sns.heatmap(pivot_cobertura, ax=ax, cmap="Blues", vmin=0, vmax=53,
            linewidths=0.3, linecolor="white", annot=True, fmt=".0f")
ax.set_title("Semanas con al menos un caso notificado por municipio y año - Magdalena")
ax.set_xlabel("Año")
ax.set_ylabel("")
plt.tight_layout()
plt.savefig(DIR_FIGURAS / "cobertura_notificacion.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
# estandarizar nombres de columnas a minusculas
df_clean.columns = df_clean.columns.str.lower()

# estandarizacion de contenido de texto a minusculas
cols_texto = ["sexo", "tip_ss", "nom_upgd","cbmte", "nom_grupo", "cod_ase"]
for col in cols_texto:
    if col in df_clean.columns:
        df_clean[col] = df_clean[col].astype(str).str.strip().str.lower().replace({"nan": pd.NA, "none": pd.NA, "": pd.NA}) #type: ignore

# fechas
cols_fecha = ["fec_not", "fec_con", "ini_sin", "fec_hos", "fec_def"]
for col in cols_fecha:
    df_clean[col] = pd.to_datetime(df_clean[col], errors="coerce")

# ocupacion
df_clean["ocupacion"] = pd.to_numeric(df_clean["ocupacion"], errors="coerce").astype("Float64")

# numericas enteras
cols_int = [
    "cod_eve", "semana", "ano", "edad", "uni_med",
    "nacionalidad",
    "cod_pais_o", "cod_dpto_o", "cod_mun_o",
    "area", "per_etn", "estrato", "sem_ges",
    "gp_discapa", "gp_desplaz", "gp_migrant", "gp_carcela", "gp_gestan",
    "gp_indigen", "gp_pobicfb", "gp_mad_com", "gp_desmovi", "gp_psiquia",
    "gp_vic_vio", "gp_otros",
    "fuente",
    "cod_pais_r", "cod_dpto_r", "cod_mun_r",
    "cod_dpto_n", "cod_mun_n",
    "tip_cas", "pac_hos", "con_fin", "ajuste", "confirmados", "estado_final_de_caso"
]
for col in cols_int:
    if col in df_clean.columns:
        df_clean[col] = pd.to_numeric(df_clean[col], errors="coerce").astype("Int64")

# gru_pob se conserva como texto para el mapeo de 4.6
df_clean["gru_pob"] = df_clean["gru_pob"].astype(str).str.strip().str.lower().replace({"nan": pd.NA, "none": pd.NA, "": pd.NA})


print(f"Filas: {df_clean.shape[0]:,} | Columnas: {df_clean.shape[1]}")
print("\nTipos de dato:")
display(df_clean.dtypes.astype(str).value_counts().reset_index().rename(columns={"index": "dtype", "count": "columnas"}))
```

## 4.3 Normalización y Validación de Edad

Se construye la variable derivada edad_anios convirtiendo edad a años según la unidad de medida registrada en uni_med. La sección 3.5 confirmó que existen registros en meses, días, horas y minutos que inflarían la distribución si se usara edad directamente.

**Tabla de conversión**

| uni_med | Unidad | Divisor |
|---|---|---|
| 1 | Años | 1 |
| 2 | Meses | 12 |
| 3 | Días | 365 |
| 4 | Horas | 8760 |
| 5 | Minutos | 525600 |

Una vez calculada edad_anios se aplica la siguiente validación: valores superiores a 120 años se imputan como nulo por corresponder a errores de digitación confirmados en la sección 3.5, pero si tras la conversión el resultado sigue siendo mayor a 120 sí es un error real. Finalmente se descartan edad y uni_med

```python
# conversion de edad a anos
divisores = {1: 1, 2: 12, 3: 365, 4: 8760, 5: 525600}
df_clean["edad_anios"] = (
    df_clean["edad"] / df_clean["uni_med"].map(divisores)
).round(2)

# validacion — valores mayores a 120 se imputan como nulo
n_invalidos = (df_clean["edad_anios"] > 120).sum()
df_clean.loc[df_clean["edad_anios"] > 120, "edad_anios"] = pd.NA
print(f"Registros con edad_anios > 120 imputados como nulo: {n_invalidos:,}")

# validacion — valores negativos
n_negativos = (df_clean["edad_anios"] < 0).sum()
df_clean.loc[df_clean["edad_anios"] < 0, "edad_anios"] = pd.NA
print(f"Registros con edad_anios negativa imputados como nulo: {n_negativos:,}")

# descarte de eda Y uni_med
df_clean = df_clean.drop(columns=["edad", "uni_med"], errors="ignore")

# verificacion
print(f"\nDistribución edad_anios:")
print(df_clean["edad_anios"].describe().round(2))
print(f"Nulos en edad_anios: {df_clean['edad_anios'].isna().sum():,}")
```

## 4.4 Marcado de Municipios Válidos

Se crea la columna auxiliar mun_valido mediante validación de cod_mun_o contra el catálogo DIVIPOLA. Los registros con código inválido se conservan en df_clean y quedan excluidos automáticamente en los análisis que requieran municipio válido, sin necesidad de intervención manual adicional.

```python
# carga del catalogo DIVIPOLA
divipola = (
    pd.read_excel(DIR_MAPPINGS / "Davipola.xlsx", dtype=str)
    .assign(cod_municipio=lambda d: d["cod_municipio"].astype(str).str.strip())
)
codigos_validos = set(pd.to_numeric(divipola["cod_municipio"], errors="coerce").dropna().astype("Int64").unique())

# construir codigo completo en df_clean
df_clean["cod_mun_completo"] = pd.to_numeric(
    df_clean["cod_dpto_o"].astype(str).str.strip() +
    df_clean["cod_mun_o"].astype(str).str.strip().str.zfill(3),
    errors="coerce"
).astype("Int64")

# marcado
df_clean["mun_valido"] = df_clean["cod_mun_completo"].isin(codigos_validos)

# verificacion
n_validos = df_clean["mun_valido"].sum()
n_invalidos = (~df_clean["mun_valido"]).sum()
print(f"Municipios válidos:   {n_validos:,} ({n_validos/len(df_clean)*100:.2f}%)")
print(f"Municipios inválidos: {n_invalidos:,} ({n_invalidos/len(df_clean)*100:.2f}%)")

df_clean
```

El 99.30% de los registros tienen municipio válido. Los registros restantes (0.70%) corresponden a casos donde el municipio de ocurrencia no pudo ser identificado al momento de la notificación o el pais de ocurrencia no era Colombia, situación documentada en la sección 3.6.

## 4.5 Carga de Tablas de Referencia

| Archivo | Columnas clave | Uso |
|---|---|---|
| Davipola.xlsx | cod_departamento, cod_municipio, municipio, departamento | Nombres de municipios y departamentos; subconjunto de 30 municipios de Magdalena para análisis departamental |
| ocupaciones.xlsx | cod_ocu, cod_antiguo, nom_ocu | Nombre de ocupación del paciente; cod_antiguo se usa como clave secundaria para registros con códigos históricos |
| aseguradoras.xls | cod_ase, nom_ase | Nombre de la EAPB/EPS |
| TablaReferencia_CIE10__1.xlsx | Codigo, Nombre | Causa básica de muerte (solo cod_eve=580) |
| paises.xlsx | codnumpais, pais | Nombre del país de ocurrencia y residencia |
| gruposetnicos.xlsx | cod_grupo, nom_grupo | Código de 3 dígitos (con cero a la izquierda) del grupo étnico registrado en gru_pob |

```python
# Cargar todas las tablas de referencia

# DIVIPOLA: departamentos y municipios
davipola  = pd.read_excel(DIR_MAPPINGS / "Davipola.xlsx", dtype=str)
dptos_dv  = davipola[["cod_departamento", "departamento"]].drop_duplicates()
muns_dv   = davipola[["cod_municipio", "municipio"]].drop_duplicates()
muns_dv_mag = (
    davipola[davipola["cod_departamento"] == "47"][["cod_municipio", "municipio"]]
    .drop_duplicates()
)

# Ocupaciones
ocupaciones = (
    pd.read_excel(DIR_MAPPINGS / "ocupaciones.xlsx", dtype=str)
    [["cod_ocu", "cod_antiguo", "nombre_vigente_ocupacion"]]
    .rename(columns={"nombre_vigente_ocupacion": "nom_ocu"})
    .drop_duplicates(subset="cod_ocu", keep="first")
)

# Aseguradoras (EAPB)
aseguradoras = (
    pd.read_excel(DIR_MAPPINGS / "aseguradoras.xls", dtype=str, engine="xlrd")
    [["cod_ase", "nom_ase"]]
    .assign(cod_ase=lambda d: d["cod_ase"].astype(str).str.strip().str.lower())
    .drop_duplicates(subset="cod_ase", keep="first")
)

# Tabla CIE-10
cie10 = (
    pd.read_excel(DIR_MAPPINGS / "TablaReferencia_CIE10__1.xlsx", dtype=str)
    [["Codigo", "Nombre"]]
    .assign(Codigo=lambda d: d["Codigo"].astype(str).str.strip().str.lower())
    .drop_duplicates(subset="Codigo", keep="first")
)

# Paises
paises = pd.read_excel(DIR_MAPPINGS / "paises.xlsx", dtype=str)
paises = paises[["codnumpais", "pais"]].drop_duplicates()

# Grupos etnicos (codigo de 3 digitos, ej. "005")
gruposetnicos = (
    pd.read_excel(DIR_MAPPINGS / "gruposetnicos.xlsx", dtype=str)
    [["cod_grupo", "nom_grupo"]]
    .assign(
        cod_grupo=lambda d: d["cod_grupo"].astype(str).str.strip().str.zfill(3),
        nom_grupo=lambda d: d["nom_grupo"].astype(str).str.strip().str.lower()
    )
    .drop_duplicates(subset="cod_grupo", keep="first")
)

print("Tablas de referencia cargadas:")
print(f"  DIVIPOLA departamentos: {len(dptos_dv)} filas")
print(f"  DIVIPOLA municipios:    {len(muns_dv)} filas")
print(f"  Municipios Magdalena:   {len(muns_dv_mag)} municipios")
print(f"  Ocupaciones:            {len(ocupaciones)} filas")
print(f"  Aseguradoras:           {len(aseguradoras)} filas")
print(f"  CIE-10:                 {len(cie10)} filas")
print(f"  Paises:                 {len(paises)} filas")
print(f"  Grupos etnicos:         {len(gruposetnicos)} filas")
```

## 4.6 Incorporación de Nombres de Referencia

Se agregan columnas de nombre legible a df_clean mediante left joins con las tablas de referencia cargadas en 4.5. El tipo left garantiza que no se pierda ningún registro del dataset principal. Se valida después de cada join que el número de filas no cambie y que los nulos generados sean esperados. El join de CIE-10 se realiza exclusivamente sobre los registros con COD_EVE=580 y el resultado se incorpora al dataset completo.

Adicionalmente, nom_grupo y gru_pob codifican la misma información (grupo étnico) en dos formatos distintos y son mutuamente excluyentes en los datos: cuando uno está poblado, el otro es nulo. Se consolida todo en nom_grupo mapeando gru_pob (con cero a la izquierda a 3 dígitos) contra el catálogo gruposetnicos, rellenando únicamente donde nom_grupo es nulo. gru_pob se descarta al finalizar la consolidación.

```python
n_antes = len(df_clean)

# departamento de ocurrencia
df_clean = df_clean.merge(
    dptos_dv.rename(columns={"cod_departamento": "cod_dpto_o", "departamento": "nom_dpto_o"})
    .assign(cod_dpto_o=lambda d: pd.to_numeric(d["cod_dpto_o"], errors="coerce").astype("Int64")),
    on="cod_dpto_o", how="left"
)

# municipio de ocurrencia
df_clean = df_clean.merge(
    muns_dv.rename(columns={"cod_municipio": "cod_mun_completo", "municipio": "nom_mun_o"})
    .assign(cod_mun_completo=lambda d: pd.to_numeric(d["cod_mun_completo"], errors="coerce").astype("Int64")),
    on="cod_mun_completo", how="left"
)

# pais de ocurrencia
df_clean = df_clean.merge(
    paises.rename(columns={"codnumpais": "cod_pais_o", "pais": "nom_pais_o"})
    .assign(cod_pais_o=lambda d: pd.to_numeric(d["cod_pais_o"], errors="coerce").astype("Int64")),
    on="cod_pais_o", how="left"
)

# ocupacion
# merge por codigo nuevo
df_clean = df_clean.merge(
    ocupaciones[["cod_ocu", "nom_ocu"]]
    .assign(cod_ocu=lambda d: pd.to_numeric(d["cod_ocu"], errors="coerce").astype("Float64"))
    .drop_duplicates(subset="cod_ocu", keep="first")
    .rename(columns={"cod_ocu": "ocupacion", "nom_ocu": "nom_ocupacion"}),
    on="ocupacion", how="left"
)
# para los que no encontraron nombre, buscar por codigo antiguo
sin_nombre = df_clean["nom_ocupacion"].isna()
df_clean.loc[sin_nombre, "nom_ocupacion"] = df_clean.loc[sin_nombre, "ocupacion"].map(
    ocupaciones
    .assign(cod_antiguo=lambda d: pd.to_numeric(d["cod_antiguo"], errors="coerce").astype("Float64"))
    .dropna(subset=["cod_antiguo"])
    .drop_duplicates(subset="cod_antiguo", keep="first")
    .set_index("cod_antiguo")["nom_ocu"]
)

# aseguradora
df_clean = df_clean.merge(
    aseguradoras.assign(cod_ase=lambda d: d["cod_ase"].astype(str).str.strip()),
    on="cod_ase", how="left"
)

# CIE-10 solo sobre cod_eve=580
reemplazos_cbmte = {
    "j15x": "j159",
    "i49x": "i499",
    "d65": "d65x",
    "8888": pd.NA,
    "9999": pd.NA,
    "none": pd.NA
}
df_clean["cbmte"] = df_clean["cbmte"].replace(reemplazos_cbmte)
df_clean = df_clean.merge(
    cie10.rename(columns={"Codigo": "cbmte", "Nombre": "nom_cbmte"}),
    on="cbmte", how="left"
)

assert len(df_clean) == n_antes, f"El merge modificó el número de filas: {len(df_clean):,} vs {n_antes:,}"
print(f"Filas conservadas: {len(df_clean):,}")
print(f"Columnas finales:  {df_clean.shape[1]}")

# nulos esperados por join
for col in ["nom_dpto_o", "nom_mun_o", "nom_pais_o", "nom_ocupacion", "nom_ase", "nom_cbmte"]:
    n = df_clean[col].isna().sum()
    print(f"  {col}: {n:,} nulos ({n/len(df_clean)*100:.2f}%)")

# consolidacion nom_grupo / gru_pob (mismo dato, dos formatos, mutuamente excluyentes)
n_nom_grupo_antes = df_clean["nom_grupo"].notna().sum()
n_gru_pob = df_clean["gru_pob"].notna().sum()

cod_gru_pob = df_clean["gru_pob"].astype(str).str.strip().str.zfill(3)
mapeo_grupo = cod_gru_pob.map(gruposetnicos.set_index("cod_grupo")["nom_grupo"])

n_sin_mapeo = df_clean["gru_pob"].notna().sum() - mapeo_grupo[df_clean["gru_pob"].notna()].notna().sum()
assert n_sin_mapeo == 0, f"Hay {n_sin_mapeo} codigos de gru_pob sin equivalencia en gruposetnicos"

sin_nom_grupo = df_clean["nom_grupo"].isna()
df_clean.loc[sin_nom_grupo, "nom_grupo"] = mapeo_grupo[sin_nom_grupo]

df_clean = df_clean.drop(columns=["gru_pob"])

n_nom_grupo_despues = df_clean["nom_grupo"].notna().sum()
assert n_nom_grupo_despues == n_nom_grupo_antes + n_gru_pob, (
    f"Conteo inesperado tras consolidar nom_grupo: {n_nom_grupo_despues:,} "
    f"vs esperado {n_nom_grupo_antes + n_gru_pob:,}"
)
print(f"\nnom_grupo consolidado: {n_nom_grupo_despues:,} no nulos "
      f"({n_nom_grupo_antes:,} originales + {n_gru_pob:,} mapeados desde gru_pob)")
```

# 6. Conclusiones y Decisiones

## 6.1 Principales Hallazgos

El dataset cubre 2007-2024 sin vacíos y conserva integridad para canal endémico. Magdalena registró 27.117 casos confirmados de dengue (210+220)

La calidad general es alta: las variables núcleo superan 97% de completitud,
los nulos estructurales (FEC_HOS, sem_ges, FEC_DEF, CBMTE) son esperados, y
el 99,71% de los registros tiene municipio válido contra DIVIPOLA.

## 6.2 Limpieza para el pipeline en clean.py

El módulo implementa un Pipeline de scikit-learn con ColumnTransformer
que reproduce, en este orden, las transformaciones validadas en la sección 4:

1. Descarte de 29 columnas: variables sin valor analítico (CONSECUTIVE, CONSECUTIVE2,
   consecutive_origen, consecutive_12,Partición, Separación ,Particion, COD_SUB, COD_PRE,
   FEC_ARC_XL, FEC_AJU, FM_*, va_sispro, CER_DEF) y variables que violan no
   transitividad (nombres redundantes con códigos, FECHA_NTO).
2. Estandarización a minúsculas de nombres de columna y de contenido de
   variables de texto.
3. Conversión de tipos: fechas a datetime, ocupacion a Float64, resto a Int64.
4. Construcción de edad_anios con divisores {1, 12, 365, 8760, 525600}
   según uni_med, imputación a nulo de valores > 120 o negativos, y descarte
   de edad y uni_med.
5. Validación geográfica: construcción de cod_mun_completo y marcado de
   mun_valido contra DIVIPOLA. Los inválidos se conservan con el flag, no
   se eliminan.
6. Enriquecimiento por left joins con DIVIPOLA, paises, ocupaciones (con
   cod_antiguo como clave secundaria), aseguradoras y CIE-10 previa
   corrección de códigos (j15x->j159, i49x->i499, d65->d65x; placeholders a nulo).

## 6.3 Limitaciones

Limitación 1 - Denominador (población en riesgo): <br>
El lineamiento metodológico del MSPS/INS estima la población en riesgo cruzando la presencia del vector por área geográfica con la transmisión de casos por área geográfica según la Tabla 5. Dado que el dato oficial de presencia vectorial desagregado por área no está disponible públicamente, este estudio infiere la presencia del vector por área geográfica directamente a partir de la transmisión histórica de casos registrada en el campo area del SIVIGILA, bajo el supuesto epidemiológicamente válido de que la ocurrencia de transmisión implica necesariamente la presencia del vector en esa misma área. Este supuesto está respaldado por los registros SIVIGILA que muestran que el 82,6% de los casos de dengue en Colombia se presentaron en cabeceras municipales y el 17,4% en centros poblados y rural disperso, patrón consistente con estudios entomológicos que confirman que la mayor probabilidad de presencia del vector se concentra en áreas cercanas a los centros urbanos (Conde et al., Biomédica, 2017), mientras que la presencia en áreas rurales ha sido documentada exclusivamente en municipios con transmisión rural activa (Ruiz-López et al., Biomédica, 2016). Bajo esta lógica, se asume presencia vectorial en cabecera municipal para los municipios con transmisión urbana confirmada, y presencia en centros poblados y rural disperso para los municipios con transmisión rural confirmada, aplicando tres escenarios: transmisión urbana y rural simultánea = población total; transmisión únicamente urbana = población de cabecera municipal; transmisión únicamente rural = población de centros poblados y rural disperso. Esta aproximación fue validada contra los datos de población en riesgo publicados oficialmente por el MSPS para el periodo 2020-2022, único periodo con información pública disponible, obteniéndose valores consistentes. Para 2023 y 2024 no existe archivo oficial publicado, por lo que se aplica la misma metodología.

Limitación 2 - Vigencia de la estratificación: <br>
La clasificación de riesgo del lineamiento MSPS/INS aplica para el periodo 2020-2023 y fue construida con datos epidemiológicos 2014-2019. Este estudio asume su validez para todo el periodo 2019-2024. La comparación con el Informe de Evento Dengue 2024 del INS evidencia cambios en la clasificación de al menos 5 municipios, entre ellos Armenia, Girardot y Cartago, que ascendieron a muy alta transmisión, confirmando que la dinámica de transmisión ha cambiado y que la estratificación puede subestimar el riesgo real en años recientes.

Limitación 3 - Clasificación clínica del dengue: <br>
SIVIGILA registra dengue sin signos de alarma, dengue con signos de alarma y dengue grave como categorías clínicas diferenciadas, así como los grupos de síntomas asociados. Sin embargo, el portal público de microdatos del SIVIGILA no expone estas variables de clasificación clínica ni el detalle sintomático en las bases de datos descargables, limitando el análisis a la ocurrencia de casos confirmados. Por tanto, este estudio no desagrega por severidad clínica ni por perfil sintomático, lo que restringe el análisis de la carga de enfermedad grave y la caracterización clínica de los casos. Para acceder a esta información se requeriría una solicitud formal de datos al INS, lo cual está fuera del alcance de este estudio.

Limitación 4 - Acceso y actualización de los datos: <br>
Los microdatos del SIVIGILA están disponibles únicamente a través de descarga manual de archivos Excel en el portal público del INS, sin acceso mediante API ni mecanismos de consulta automatizada. Esto implica que el periodo de análisis está limitado a los datos disponibles al momento de la descarga, siendo 2024 el último año con información consolidada accesible. No es posible incorporar datos más recientes ni mantener el sistema actualizado de forma automática sin una nueva descarga manual, lo que limita la capacidad del dashboard para reflejar la situación epidemiológica en tiempo real.
Para abordar esta limitación, el sistema contempla un mecanismo de actualización de datos que permite incorporar nuevos periodos sin modificar el código. Esto es viable porque los microdatos del SIVIGILA mantienen una estructura de columnas consistente entre años, lo que permite que el pipeline de procesamiento opere de forma estandarizada independientemente del año incorporado. De esta manera, aunque la actualización no es automática, el sistema puede extenderse a nuevos periodos con una intervención mínima, preservando la reproducibilidad y trazabilidad del proceso.
