# 1. Introduccion

## 1.1 Contexto del Dataset

Este dataset proviene del Sistema de Vigilancia de Salud Publica (SIVIGILA), que recopila y analiza datos sobre eventos que afectan la salud de la poblacion colombiana: enfermedades infecciosas, cronicas, factores de riesgo y otros sucesos de importancia en salud publica.

El dengue es un evento de notificacion obligatoria en Colombia desde la reemergencia de los anos setenta. La notificacion se realiza de forma individual por las Unidades Primarias Generadoras de Datos (UPGD) al sistema SIVIGILA. Se manejan 3 eventos de notificacion relacionados al dengue: dengue (210), dengue grave (220) y mortalidad por dengue (580).

## 1.2 Descripcion del Conjunto de Datos

Los datos se obtienen del portal oficial del INS - SIVIGILA:
https://portalsivigila.ins.gov.co/Paginas/Buscador.aspx

Cada fila representa un caso individual notificado con informacion demografica, geografica, clinica y administrativa.

> Diccionario oficial: https://www.ins.gov.co/BibliotecaDigital/diccionario-xls-.pdf

## 3.3 Distribucion de COD_EVE

El campo COD_EVE define el tipo de evento notificado y determina que analisis aplica a cada registro: incidencia (210, 220) o mortalidad (580). Esta seccion examina la proporcion de cada evento.

```python
conteo_cod_eve = (
    df_raw["COD_EVE"].astype(str).str.strip()
    .value_counts().reset_index()
)
conteo_cod_eve.columns = ["COD_EVE", "conteo"]
conteo_cod_eve["pct"] = (conteo_cod_eve["conteo"] / len(df_raw) * 100).round(2)

etiquetas = {"210": "Dengue (210)", "220": "Dengue grave (220)", "580": "Mortalidad (580)"}
conteo_cod_eve["label"] = conteo_cod_eve["COD_EVE"].map(etiquetas)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

wedges, _, autotexts = axes[0].pie(
    conteo_cod_eve["conteo"],
    labels=None,
    autopct="%1.1f%%",
    colors=[INS_AZUL, INS_VERDE, INS_ROJO],
    startangle=90,
    pctdistance=0.75,
    wedgeprops={"edgecolor": "white", "linewidth": 1.5}
)
for at in autotexts:
    at.set_color("white")
    at.set_fontsize(9)
axes[0].legend(
    wedges, conteo_cod_eve["label"],
    loc="lower center", bbox_to_anchor=(0.5, -0.12),
    fontsize=9, frameon=False
)
axes[0].set_title("Distribución por tipo de evento", fontsize=11)

detalle = conteo_cod_eve[conteo_cod_eve["COD_EVE"].isin(["220", "580"])]
axes[1].barh(detalle["label"], detalle["conteo"], color=[INS_VERDE, INS_ROJO], edgecolor="white")
for i, (val, pct) in enumerate(zip(detalle["conteo"], detalle["pct"])):
    axes[1].text(val + 200, i, f"{val:,} ({pct}%)", va="center", fontsize=9)
axes[1].set_xlabel("Registros")
axes[1].set_title("Detalle: dengue grave y mortalidad", fontsize=11)

plt.suptitle("Distibución de COD_EVE (210, 220, 580)", fontsize=12)
plt.tight_layout()
plt.savefig(DIR_FIGURAS / "distribucion_cod_eve.png", dpi=150, bbox_inches="tight")
plt.show()
```

El dengue (210) concentra el 98.7% de los registros (751,114 casos), mientras que el dengue grave (220) representa el 1.21% (9,227 casos) y la mortalidad por dengue (580) el 0.1% (753 casos). Esta proporcion es coherente con la piramide de severidad esperada: la gran mayoria de infecciones cursan como dengue sin senales de alarma, una fraccion menor evoluciona a dengue grave y solo un porcentaje minimo resulta en muerte. Los 753 registros de mortalidad, aunque minoritarios, son suficientes para el analisis de la seccion 5.9.

## 3.4 Clasificación y Estado Final del Caso

Los campos TIP_CAS, AJUSTE, confirmados y Estado_final_de_caso determinan conjuntamente el ciclo de vida de un caso en SIVIGILA, desde su clasificación inicial hasta su estado definitivo tras el proceso de ajuste. El protocolo INS indica que deben excluirse los registros con AJUSTE = 6 (descartado) y AJUSTE = D (error de digitación) antes de cualquier análisis. ## 3.6 Cobertura de Notificacion por Municipio

Numero de semanas epidemiologicas con al menos un caso notificado por municipio y ano. Un valor de 0 indica que el municipio no reporto ningun caso durante ese ano completo.

```python
_df = df_raw.copy()
_df["ANO"]    = pd.to_numeric(_df["ANO"],    errors="coerce")
_df["SEMANA"] = pd.to_numeric(_df["SEMANA"], errors="coerce")
_df = (
    _df[_df["COD_DPTO_O"].astype(str).str.lstrip("0") == "47"]
    .dropna(subset=["COD_MUN_O", "ANO", "SEMANA"])
)

anos_disponibles = sorted(_df["ANO"].unique().astype(int))

semanas_con_casos = (
    _df.groupby(["COD_MUN_O", "ANO", "SEMANA"])
    .size()
    .reset_index(name="casos")
    .groupby(["COD_MUN_O", "ANO"])
    .size()
    .reset_index(name="sem_con_casos")
)

pivot_cobertura = semanas_con_casos.pivot(
    index="COD_MUN_O", columns="ANO", values="sem_con_casos"
).fillna(0)

_muns = (
    pd.read_excel(DIR_MAPPINGS / "Davipola.xlsx", dtype=str)
    .assign(cod_municipio=lambda d: "47" + d["cod_municipio"].astype(str).str.strip().str[-3:])
    .query("cod_departamento.str.strip().str.lstrip('0') == '47'", engine="python")
    [["cod_municipio", "municipio"]]
    .drop_duplicates()
)

pivot_cobertura = pivot_cobertura.reset_index()
pivot_cobertura["COD_MUN_O"] = "47" + pivot_cobertura["COD_MUN_O"].astype(str).str.strip().str.zfill(3)
pivot_cobertura = pivot_cobertura.merge(_muns, left_on="COD_MUN_O", right_on="cod_municipio", how="left")
pivot_cobertura["municipio"] = pivot_cobertura["municipio"].fillna(pivot_cobertura["COD_MUN_O"])
pivot_cobertura = (
    pivot_cobertura
    .set_index("municipio")
    .drop(columns=["COD_MUN_O", "cod_municipio"], errors="ignore")
    .sort_values(anos_disponibles[-1], ascending=False)
)

fig, ax = plt.subplots(figsize=(len(anos_disponibles) + 3, len(pivot_cobertura) * 0.38 + 2))
sns.heatmap(pivot_cobertura, ax=ax, cmap="Blues", vmin=0, vmax=53,
            linewidths=0.3, linecolor="white", annot=True, fmt=".0f")
ax.set_title("Semanas con al menos un caso notificado por municipio y año — Magdalena")
ax.set_xlabel("Año")
ax.set_ylabel("")
plt.tight_layout()
plt.savefig(DIR_FIGURAS / "cobertura_notificacion.png", dpi=150, bbox_inches="tight")
plt.show()
```

## 4.1 Transformaciones Aplicadas

| Paso | Accion | Justificacion |
|---|---|---|
| 1 | Filtro AJUSTE 6/D | La seccion 3.4 confirmo 0 registros con esos valores - se omite pero se documenta |

```python
df_clean = df_raw.copy()

# Paso 1: Estandarizar columnas y contenido de texto a minusculas
df_clean.columns = df_clean.columns.str.lower().str.strip()
cols_texto = df_clean.select_dtypes(include="object").columns
for col in cols_texto:
    df_clean[col] = df_clean[col].str.lower().str.strip()
print(f"Columnas estandarizadas: {df_clean.shape[1]}")
print(f"Columnas de texto normalizadas a minusculas: {len(cols_texto)}")

# Paso 2: AJUSTE 6/D
# La seccion 3.4 confirmo que no existen registros con AJUSTE=6 o AJUSTE=D.
# El filtro se omite pero se deja documentado para reproducibilidad.
n_descartados = df_clean["ajuste"].isin(["6", "d"]).sum()
print(f"\nRegistros con AJUSTE 6 o D: {n_descartados:,}  (confirmado en seccion 3.4 - sin cambios)")

# Paso 3: Conversion de tipos
cols_fecha = ["fec_not", "ini_sin", "fec_hos", "fec_def", "fec_aju", "fec_con", "fecha_nto"]
for col in cols_fecha:
    if col in df_clean.columns:
        df_clean[col] = pd.to_datetime(df_clean[col], errors="coerce")

cols_int = ["semana", "ano", "cod_eve", "pac_hos", "con_fin", "tip_cas", "fuente",
            "per_etn", "area", "uni_med", "ajuste", "confirmados", "estado_final_de_caso",
            "nacionalidad", "cod_pais_o", "cod_dpto_o", "cod_mun_o",
            "cod_pais_r", "cod_dpto_r", "cod_mun_r", "cod_dpto_n", "cod_mun_n",
            "ocupacion", "gp_discapa", "gp_desplaz", "gp_migrant", "gp_carcela",
            "gp_gestan", "gp_indigen", "gp_pobicfb", "gp_mad_com", "gp_desmovi",
            "gp_psiquia", "gp_vic_vio", "gp_otros"]
for col in cols_int:
    if col in df_clean.columns:
        df_clean[col] = pd.to_numeric(df_clean[col], errors="coerce")

cols_int64 = ["edad", "semana", "ano", "estrato", "sem_ges"]
for col in cols_int64:
    if col in df_clean.columns:
        df_clean[col] = pd.to_numeric(df_clean[col], errors="coerce").astype("Int64")

print(f"Fechas convertidas: {cols_fecha}")
print(f"Numericos convertidos: {[c for c in cols_int if c in df_clean.columns]}")

# Paso 4: Normalizar EDAD a anos
df_clean["edad_anios"] = df_clean["edad"].astype(float)
uni = df_clean["uni_med"].astype(str).str.strip()
df_clean.loc[uni == "2", "edad_anios"] = df_clean["edad_anios"] / 12
df_clean.loc[uni == "3", "edad_anios"] = df_clean["edad_anios"] / 365
df_clean.loc[uni == "4", "edad_anios"] = df_clean["edad_anios"] / 8760
df_clean["edad_anios"] = df_clean["edad_anios"].round(2)
print(f"\nedad_anios - rango: {df_clean['edad_anios'].min():.1f} a {df_clean['edad_anios'].max():.1f} anos")

# Paso 5: Validar cod_mun_o contra catalogo DIVIPOLA (solo registros de Colombia)
_davipola = pd.read_excel(DIR_MAPPINGS / "Davipola.xlsx", dtype=str)
municipios_validos = set(_davipola["cod_municipio"].astype(str).str.strip().str[-3:])

df_colombia = df_clean[df_clean["pais_ocurrencia"].astype(str).str.strip() == "colombia"]
cod_mun_limpio = df_colombia["cod_mun_o"].astype(str).str.strip().str.zfill(3)
n_invalidos = (~cod_mun_limpio.isin(municipios_validos)).sum()

print(f"\nRegistros colombianos con cod_mun_o invalido: {n_invalidos:,}")
display(
    df_colombia[~cod_mun_limpio.isin(municipios_validos)]
    [["cod_mun_o", "municipio_ocurrencia"]]
    .value_counts()
    .reset_index()
    .rename(columns={0: "conteo"})
)

mask_validos = (
    (df_clean["pais_ocurrencia"].astype(str).str.strip() != "colombia") |
    (df_clean["cod_mun_o"].astype(str).str.strip().str.zfill(3).isin(municipios_validos))
)
df_clean = df_clean[mask_validos].copy()
print(f"df_clean tras excluir municipios invalidos: {len(df_clean):,} registros")
```

```python
# Cargar todas las tablas de referencia 

# DIVIPOLA: departamentos y municipios
davipola  = pd.read_excel(DIR_MAPPINGS / "Davipola.xlsx", dtype=str)
dptos_dv  = davipola[["cod_departamento", "departamento"]].drop_duplicates()
muns_dv   = davipola[["cod_municipio", "municipio"]].drop_duplicates()
muns_dv_mag = (
    davipola[davipola["cod_departamento"] == "47"][["cod_municipio", "municipio"]]
    .drop_duplicates()
)

# Ocupaciones
ocupaciones = pd.read_excel(DIR_MAPPINGS / "ocupaciones.xlsx", dtype=str)
ocupaciones = ocupaciones[["cod_ocu", "nom_ocu"]].drop_duplicates()

# Aseguradoras (EAPB)
aseguradoras = pd.read_excel(DIR_MAPPINGS / "aseguradoras.xls", dtype=str, engine="xlrd")
aseguradoras = aseguradoras[["cod_ase", "nom_ase"]].drop_duplicates()

# Tabla CIE-10
cie10 = pd.read_excel(DIR_MAPPINGS / "TablaReferencia_CIE10__1.xlsx", dtype=str)
cie10 = cie10[["Codigo", "Nombre"]].drop_duplicates()

# Paises
paises = pd.read_excel(DIR_MAPPINGS / "paises.xlsx", dtype=str)
paises = paises[["codnumpais", "pais"]].drop_duplicates()

print("Tablas de referencia cargadas:")
print(f"  DIVIPOLA departamentos: {len(dptos_dv)} filas")
print(f"  DIVIPOLA municipios:    {len(muns_dv)} filas")
print(f"  Municipios Magdalena:   {len(muns_dv_mag)} municipios")
print(f"  Ocupaciones:            {len(ocupaciones)} filas")
print(f"  Aseguradoras:           {len(aseguradoras)} filas")
print(f"  CIE-10:                 {len(cie10)} filas")
print(f"  Paises:                 {len(paises)} filas")
```

## 4.4 Enriquecimiento del Dataset

Se agregan columnas de nombre legible a df_clean mediante left joins con las tablas de referencia. El tipo left garantiza que no se pierda ningun registro del dataset principal. Se valida despues de cada join que el numero de filas no cambie y que los nulos generados sean esperados.

El join de CIE-10 se hace solo sobre los registros con cod_eve = 580 y el resultado se incorpora al dataset completo.

```python
# 1. municipio y departamento de ocurrencia
lookup_mun = davipola[["cod_municipio", "municipio", "departamento"]].drop_duplicates("cod_municipio")
n0 = len(df_clean)
df_clean = df_clean.merge(
    lookup_mun.rename(columns={"municipio": "nom_municipio_o", "departamento": "nom_dpto_o"}),
    left_on="cod_mun_o", right_on="cod_municipio", how="left"
).drop(columns="cod_municipio")
print(f"Filas antes: {n0:,}  despues: {len(df_clean):,}")
print(f"Nulos nom_municipio_o: {df_clean['nom_municipio_o'].isna().sum():,}")
print(f"Nulos nom_dpto_o: {df_clean['nom_dpto_o'].isna().sum():,}")

# 2. ocupacion
n0 = len(df_clean)
df_clean = df_clean.merge(ocupaciones, left_on="ocupacion", right_on="cod_ocu", how="left").drop(columns="cod_ocu")
print(f"\nFilas antes: {n0:,}  despues: {len(df_clean):,}")
print(f"Nulos nom_ocu: {df_clean['nom_ocu'].isna().sum():,}")

# 3. aseguradora
n0 = len(df_clean)
df_clean = df_clean.merge(aseguradoras, on="cod_ase", how="left")
print(f"\nFilas antes: {n0:,}  despues: {len(df_clean):,}")
print(f"Nulos nom_ase: {df_clean['nom_ase'].isna().sum():,}")

# 4. pais de ocurrencia
n0 = len(df_clean)
df_clean = df_clean.merge(
    paises.rename(columns={"pais": "nom_pais_o"}),
    left_on="cod_pais_o", right_on="codnumpais", how="left"
).drop(columns="codnumpais")
print(f"\nFilas antes: {n0:,}  despues: {len(df_clean):,}")
print(f"Nulos nom_pais_o: {df_clean['nom_pais_o'].isna().sum():,}")

# 5. pais de residencia
n0 = len(df_clean)
df_clean = df_clean.merge(
    paises.rename(columns={"pais": "nom_pais_r"}),
    left_on="cod_pais_r", right_on="codnumpais", how="left"
).drop(columns="codnumpais")
print(f"\nFilas antes: {n0:,}  despues: {len(df_clean):,}")
print(f"Nulos nom_pais_r: {df_clean['nom_pais_r'].isna().sum():,}")

# 6. CIE-10 (solo cod_eve = 580)
n0 = len(df_clean)
idx_580 = df_clean[df_clean["cod_eve"] == "580"].index
df_clean["nom_causa_muerte"] = pd.NA
df_clean.loc[idx_580, "nom_causa_muerte"] = (
    df_clean.loc[idx_580]
    .merge(cie10.rename(columns={"Nombre": "nom_causa_muerte"}),
           left_on="cbmte", right_on="Codigo", how="left")
    ["nom_causa_muerte"].values
)
print(f"\nFilas antes: {n0:,}  despues: {len(df_clean):,}")
print(f"Nulos nom_causa_muerte: {df_clean['nom_causa_muerte'].isna().sum():,}")
```
