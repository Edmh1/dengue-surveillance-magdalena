# 1. Introduccion

## 1.1 Contexto del Dataset

Este dataset proviene del Sistema de Vigilancia de Salud Publica (SIVIGILA), que recopila y analiza datos sobre eventos que afectan la salud de la poblacion colombiana: enfermedades infecciosas, cronicas, factores de riesgo y otros sucesos de importancia en salud publica.

El dengue es un evento de notificacion obligatoria en Colombia desde la reemergencia de los años setenta. La notificacion se realiza de forma individual por las Unidades Primarias Generadoras de Datos (UPGD) al sistema SIVIGILA. Se manejan 3 eventos de notificacion relacionados al dengue: dengue (210), dengue grave (220) y mortalidad por dengue (580).

El período de análisis comprende los cinco años más recientes disponibles en fuentes abiertas del INS (2019–2024). Este rango responde al requerimiento metodológico del canal endémico, componente solicitado dentro del alcance del proyecto, que demanda como mínimo cinco años de historia epidemiológica para construir una línea de base estable sobre el año de análisis. Como consecuencia de este criterio, el período incluye el ciclo epidémico de 2019, uno de los más relevantes registrados en Colombia y la Región de las Américas.

[1.1.2. Situación epidemiológica en América](https://www.ins.gov.co/buscador-eventos/Lineamientos/Pro_Dengue.pdf)

## 1.2 Descripcion del Conjunto de Datos

https://portalsivigila.ins.gov.co/Paginas/Buscador.aspx

## 3.3 Distribucion de COD_EVE

El campo COD_EVE define el tipo de evento notificado y determina que analisis aplica a cada registro: incidencia (210, 220) o mortalidad (580). Esta seccion examina la proporcion de cada evento.

```python
conteo_cod_eve = (
    df_raw["COD_EVE"].astype(str).str.strip()
    .value_counts().reset_index()
)
conteo_cod_eve.columns = ["COD_EVE", "conteo"]
conteo_cod_eve["pct"] = (conteo_cod_eve["conteo"] / len(df_raw) * 100).round(2)

etiquetas = {"210": "Dengue (210)", "220": "Dengue grave (220)", "580": "Mortalidad (580)"}
conteo_cod_eve["label"] = conteo_cod_eve["COD_EVE"].map(etiquetas)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

wedges, _, autotexts = axes[0].pie(
    conteo_cod_eve["conteo"],
    labels=None,
    autopct="%1.1f%%",
    colors=[INS_AZUL, INS_VERDE, INS_ROJO],
    startangle=90,
    pctdistance=0.75,
    wedgeprops={"edgecolor": "white", "linewidth": 1.5}
)
for at in autotexts:
    at.set_color("white")
    at.set_fontsize(9)
axes[0].legend(
    wedges, conteo_cod_eve["label"],
    loc="lower center", bbox_to_anchor=(0.5, -0.12),
    fontsize=9, frameon=False
)
axes[0].set_title("Distribución por tipo de evento", fontsize=11)

detalle = conteo_cod_eve[conteo_cod_eve["COD_EVE"].isin(["220", "580"])]
axes[1].barh(detalle["label"], detalle["conteo"], color=[INS_VERDE, INS_ROJO], edgecolor="white")
for i, (val, pct) in enumerate(zip(detalle["conteo"], detalle["pct"])):
    axes[1].text(val + 200, i, f"{val:,} ({pct}%)", va="center", fontsize=9)
axes[1].set_xlabel("Registros")
axes[1].set_title("Detalle: dengue grave y mortalidad", fontsize=11)

plt.suptitle("Distibución de COD_EVE (210, 220, 580)", fontsize=12)
plt.tight_layout()
plt.savefig(DIR_FIGURAS / "distribucion_cod_eve.png", dpi=150, bbox_inches="tight")
plt.show()
```

## 3.4 Clasificación y Estado Final del Caso

Los campos TIP_CAS, AJUSTE, confirmados y Estado_final_de_caso determinan conjuntamente el ciclo de vida de un caso en SIVIGILA, desde su clasificación inicial hasta su estado definitivo tras el proceso de ajuste. El protocolo INS indica que deben excluirse los registros con AJUSTE = 6 (descartado) y AJUSTE = D (error de digitación) antes de cualquier análisis. Esta subsección documenta la distribución real de los tres campos y detecta valores inesperados o inconsistencias entre ellos.

## 3.6 Cobertura de Notificacion por Municipio

Numero de semanas epidemiologicas con al menos un caso notificado por municipio y ano. Un valor de 0 indica que el municipio no reporto ningun caso durante ese ano completo.

```python
_df = df_raw.copy()
_df["ANO"]    = pd.to_numeric(_df["ANO"],    errors="coerce")
_df["SEMANA"] = pd.to_numeric(_df["SEMANA"], errors="coerce")
_df = (
    _df[_df["COD_DPTO_O"].astype(str).str.lstrip("0") == "47"]
    .dropna(subset=["COD_MUN_O", "ANO", "SEMANA"])
)

anos_disponibles = sorted(_df["ANO"].unique().astype(int))

semanas_con_casos = (
    _df.groupby(["COD_MUN_O", "ANO", "SEMANA"])
    .size()
    .reset_index(name="casos")
    .groupby(["COD_MUN_O", "ANO"])
    .size()
    .reset_index(name="sem_con_casos")
)

pivot_cobertura = semanas_con_casos.pivot(
    index="COD_MUN_O", columns="ANO", values="sem_con_casos"
).fillna(0)

_muns = (
    pd.read_excel(DIR_MAPPINGS / "Davipola.xlsx", dtype=str)
    .assign(cod_municipio=lambda d: "47" + d["cod_municipio"].astype(str).str.strip().str[-3:])
    .query("cod_departamento.str.strip().str.lstrip('0') == '47'", engine="python")
    [["cod_municipio", "municipio"]]
    .drop_duplicates()
)

pivot_cobertura = pivot_cobertura.reset_index()
pivot_cobertura["COD_MUN_O"] = "47" + pivot_cobertura["COD_MUN_O"].astype(str).str.strip().str.zfill(3)
pivot_cobertura = pivot_cobertura.merge(_muns, left_on="COD_MUN_O", right_on="cod_municipio", how="left")
pivot_cobertura["municipio"] = pivot_cobertura["municipio"].fillna(pivot_cobertura["COD_MUN_O"])
pivot_cobertura = (
    pivot_cobertura
    .set_index("municipio")
    .drop(columns=["COD_MUN_O", "cod_municipio"], errors="ignore")
    .sort_values(anos_disponibles[-1], ascending=False)
)

fig, ax = plt.subplots(figsize=(len(anos_disponibles) + 3, len(pivot_cobertura) * 0.38 + 2))
sns.heatmap(pivot_cobertura, ax=ax, cmap="Blues", vmin=0, vmax=53,
            linewidths=0.3, linecolor="white", annot=True, fmt=".0f")
ax.set_title("Semanas con al menos un caso notificado por municipio y año - Magdalena")
ax.set_xlabel("Año")
ax.set_ylabel("")
plt.tight_layout()
plt.savefig(DIR_FIGURAS / "cobertura_notificacion.png", dpi=150, bbox_inches="tight")
plt.show()
```

Las variables SEXO, TIP_SS, Nom_upgd, CBMTE, nom_grupo, cod_ase se estandarizarán a minusculas sus columnas y valores

## 4.3 Normalización y Validación de Edad

Se construye la variable derivada edad_anios convirtiendo edad a años según la unidad de medida registrada en uni_med. La sección 3.5 confirmó que existen registros en meses, días, horas y minutos que inflarían la distribución si se usara edad directamente.

**Tabla de conversión**

| uni_med | Unidad | Divisor |
|---|---|---|
| 1 | Años | 1 |
| 2 | Meses | 12 |
| 3 | Días | 365 |
| 4 | Horas | 8760 |
| 5 | Minutos | 525600 |

Una vez calculada edad_anios se aplica la siguiente validación: valores superiores a 120 años se imputan como nulo por corresponder a errores de digitación confirmados en la sección 3.5, pero si tras la conversión el resultado sigue siendo mayor a 120 sí es un error real. Finalmente se descartan edad y uni_med

```python
# conversion de edad a anos
divisores = {1: 1, 2: 12, 3: 365, 4: 8760, 5: 525600}
df_clean["edad_anios"] = (
    df_clean["edad"] / df_clean["uni_med"].map(divisores)
).round(2)

# validacion — valores mayores a 120 se imputan como nulo
n_invalidos = (df_clean["edad_anios"] > 120).sum()
df_clean.loc[df_clean["edad_anios"] > 120, "edad_anios"] = pd.NA
print(f"Registros con edad_anios > 120 imputados como nulo: {n_invalidos:,}")

# validacion — valores negativos
n_negativos = (df_clean["edad_anios"] < 0).sum()
df_clean.loc[df_clean["edad_anios"] < 0, "edad_anios"] = pd.NA
print(f"Registros con edad_anios negativa imputados como nulo: {n_negativos:,}")

# descarte de eda Y uni_med
df_clean = df_clean.drop(columns=["edad", "uni_med"], errors="ignore")

# verificacion
print(f"\nDistribución edad_anios:")
print(df_clean["edad_anios"].describe().round(2))
print(f"Nulos en edad_anios: {df_clean['edad_anios'].isna().sum():,}")
```

## 4.5 Carga de Tablas de Referencia

| Archivo | Columnas clave | Uso |
|---|---|---|
| Davipola.xlsx | cod_departamento, cod_municipio, municipio, departamento | Nombres de municipios y departamentos |
| ocupaciones.xlsx | cod_ocu, nom_ocu | Nombre de ocupacion del paciente |
| aseguradoras.xls | cod_ase, nom_ase | Nombre de la EAPB/EPS |
| TablaReferencia_CIE10__1.xlsx | Codigo, Nombre | Causa basica de muerte (solo cod_eve=580) |
| paises.xlsx | codnumpais, pais | Nombre del pais de ocurrencia y residencia |

```python
# Cargar todas las tablas de referencia 

# DIVIPOLA: departamentos y municipios
davipola  = pd.read_excel(DIR_MAPPINGS / "Davipola.xlsx", dtype=str)
dptos_dv  = davipola[["cod_departamento", "departamento"]].drop_duplicates()
muns_dv   = davipola[["cod_municipio", "municipio"]].drop_duplicates()
muns_dv_mag = (
    davipola[davipola["cod_departamento"] == "47"][["cod_municipio", "municipio"]]
    .drop_duplicates()
)

# Ocupaciones
ocupaciones = (
    pd.read_excel(DIR_MAPPINGS / "ocupaciones.xlsx", dtype=str)
    [["cod_ocu", "cod_antiguo", "nombre_vigente_ocupacion"]]
    .rename(columns={"nombre_vigente_ocupacion": "nom_ocu"})
    .drop_duplicates(subset="cod_ocu", keep="first")
)

# Aseguradoras (EAPB)
aseguradoras = (
    pd.read_excel(DIR_MAPPINGS / "aseguradoras.xls", dtype=str, engine="xlrd")
    [["cod_ase", "nom_ase"]]
    .assign(cod_ase=lambda d: d["cod_ase"].astype(str).str.strip().str.lower())
    .drop_duplicates(subset="cod_ase", keep="first")
)

# Tabla CIE-10
cie10 = (
    pd.read_excel(DIR_MAPPINGS / "TablaReferencia_CIE10__1.xlsx", dtype=str)
    [["Codigo", "Nombre"]]
    .assign(Codigo=lambda d: d["Codigo"].astype(str).str.strip().str.lower())
    .drop_duplicates(subset="Codigo", keep="first")
)

# Paises
paises = pd.read_excel(DIR_MAPPINGS / "paises.xlsx", dtype=str)
paises = paises[["codnumpais", "pais"]].drop_duplicates()

print("Tablas de referencia cargadas:")
print(f"  DIVIPOLA departamentos: {len(dptos_dv)} filas")
print(f"  DIVIPOLA municipios:    {len(muns_dv)} filas")
print(f"  Municipios Magdalena:   {len(muns_dv_mag)} municipios")
print(f"  Ocupaciones:            {len(ocupaciones)} filas")
print(f"  Aseguradoras:           {len(aseguradoras)} filas")
print(f"  CIE-10:                 {len(cie10)} filas")
print(f"  Paises:                 {len(paises)} filas")
```

## 4.6 Incorporación de Nombres de Referencia

Se agregan columnas de nombre legible a df_clean mediante left joins con las tablas de referencia cargadas en 4.5. El tipo left garantiza que no se pierda ningún registro del dataset principal. Se valida después de cada join que el número de filas no cambie y que los nulos generados sean esperados. El join de CIE-10 se realiza exclusivamente sobre los registros con COD_EVE=580 y el resultado se incorpora al dataset completo.

```python
n_antes = len(df_clean)

# departamento de ocurrencia
df_clean = df_clean.merge(
    dptos_dv.rename(columns={"cod_departamento": "cod_dpto_o", "departamento": "nom_dpto_o"})
    .assign(cod_dpto_o=lambda d: pd.to_numeric(d["cod_dpto_o"], errors="coerce").astype("Int64")),
    on="cod_dpto_o", how="left"
)

# municipio de ocurrencia
df_clean = df_clean.merge(
    muns_dv.rename(columns={"cod_municipio": "cod_mun_completo", "municipio": "nom_mun_o"})
    .assign(cod_mun_completo=lambda d: pd.to_numeric(d["cod_mun_completo"], errors="coerce").astype("Int64")),
    on="cod_mun_completo", how="left"
)

# pais de ocurrencia
df_clean = df_clean.merge(
    paises.rename(columns={"codnumpais": "cod_pais_o", "pais": "nom_pais_o"})
    .assign(cod_pais_o=lambda d: pd.to_numeric(d["cod_pais_o"], errors="coerce").astype("Int64")),
    on="cod_pais_o", how="left"
)

# ocupacion
# merge por codigo nuevo
df_clean = df_clean.merge(
    ocupaciones[["cod_ocu", "nom_ocu"]]
    .assign(cod_ocu=lambda d: pd.to_numeric(d["cod_ocu"], errors="coerce").astype("Float64"))
    .drop_duplicates(subset="cod_ocu", keep="first")
    .rename(columns={"cod_ocu": "ocupacion", "nom_ocu": "nom_ocupacion"}),
    on="ocupacion", how="left"
)
# para los que no encontraron nombre, buscar por codigo antiguo
sin_nombre = df_clean["nom_ocupacion"].isna()
df_clean.loc[sin_nombre, "nom_ocupacion"] = df_clean.loc[sin_nombre, "ocupacion"].map(
    ocupaciones
    .assign(cod_antiguo=lambda d: pd.to_numeric(d["cod_antiguo"], errors="coerce").astype("Float64"))
    .dropna(subset=["cod_antiguo"])
    .drop_duplicates(subset="cod_antiguo", keep="first")
    .set_index("cod_antiguo")["nom_ocu"]
)

# aseguradora
df_clean = df_clean.merge(
    aseguradoras.assign(cod_ase=lambda d: d["cod_ase"].astype(str).str.strip()),
    on="cod_ase", how="left"
)

# CIE-10 solo sobre cod_eve=580
reemplazos_cbmte = {
    "j15x": "j159",
    "i49x": "i499",
    "8888": pd.NA,
    "9999": pd.NA,
    "none": pd.NA
}
df_clean["cbmte"] = df_clean["cbmte"].replace(reemplazos_cbmte)
df_clean = df_clean.merge(
    cie10.rename(columns={"Codigo": "cbmte", "Nombre": "nom_cbmte"}),
    on="cbmte", how="left"
)

# validacion
assert len(df_clean) == n_antes, f"El merge modificó el número de filas: {len(df_clean):,} vs {n_antes:,}"
print(f"Filas conservadas: {len(df_clean):,}")
print(f"Columnas finales:  {df_clean.shape[1]}")

# nulos esperados por join
for col in ["nom_dpto_o", "nom_mun_o", "nom_pais_o", "nom_ocupacion", "nom_ase", "nom_cbmte"]:
    n = df_clean[col].isna().sum()
    print(f"  {col}: {n:,} nulos ({n/len(df_clean)*100:.2f}%)")
```

## 5.5 Panorama General Nacional

## 5.9 Analisis Clinico — Magdalena

Caracterizacion clinica de los casos: tipo de clasificacion (tip_cas), hospitalizacion (pac_hos), condicion final (con_fin) y distribucion por tipo de evento (210 vs 220).

```python
fig, axes = plt.subplots(2, 2, figsize=(16, 12))
axes = axes.flatten()

# 1: Clasificación inicial vs. final ───────────────────────────────────────
TIP_CAS_LABELS = {
    "1": "Sospechoso", "2": "Probable", "3": "Conf. laboratorio",
    "4": "Conf. clínica", "5": "Conf. nexo epi."
}
tip_ini = df_mag["tip_cas"].astype(str).map(TIP_CAS_LABELS).fillna(df_mag["tip_cas"]).value_counts()
tip_fin = df_mag["estado_final_de_caso"].astype(str).value_counts()

for i, (serie, titulo, color) in enumerate([(tip_ini, "Clasificación inicial (TIP_CAS)", INS_AZUL),
                                              (tip_fin, "Clasificación final", INS_ROJO)]):
    serie_s = serie.sort_values(ascending=True)
    axes[i].barh(serie_s.index.astype(str), serie_s.values, color=color)
    axes[i].set_title(titulo)
    axes[i].set_xlabel("Registros")

cambios     = (df_mag["tip_cas"].astype(str) != df_mag["estado_final_de_caso"].astype(str)).sum()
pct_cambios = cambios / len(df_mag) * 100
print(f"Registros que cambian de clasificación: {cambios:,} ({pct_cambios:.1f}%)")

# 2: Hospitalización por tipo de caso ──────────────────────────────────────
dengue_mag = df_mag[df_mag["COD_EVE"].isin(["210", "220"])].copy()
dengue_mag["hospitalizado"] = dengue_mag["pac_hos"].astype(str).str.strip() == "1"
dengue_mag["tipo_caso"]     = dengue_mag["COD_EVE"].map({"210": "Dengue (210)", "220": "Dengue grave (220)"})

hosp_pct = (
    dengue_mag.groupby(["ANO", "tipo_caso"])["hospitalizado"]
    .mean().mul(100).reset_index()
)
hosp_pct.columns = ["ANO", "tipo_caso", "pct"]
pivot_hosp = hosp_pct.pivot(index="ANO", columns="tipo_caso", values="pct").fillna(0)
pivot_hosp.plot(kind="bar", ax=axes[2], color=[INS_AZUL, INS_ROJO], edgecolor="white")
axes[2].set_title("% hospitalizados por año y tipo de caso")
axes[2].yaxis.set_major_formatter(mticker.PercentFormatter())
axes[2].legend(fontsize=9)

# 3: Condición final ───────────
CON_FIN_LABELS = {"0": "No sabe", "1": "Vivo", "2": "Muerto"}
cf = df_mag.copy()
cf["condicion"]  = cf["con_fin"].astype(str).str.strip().map(CON_FIN_LABELS).fillna(cf["con_fin"])
cf["tipo_caso"]  = cf["COD_EVE"].map({"210": "Dengue (210)", "220": "Dengue grave (220)", "580": "Mortalidad (580)"}).fillna(cf["COD_EVE"])

pivot_cf = cf.groupby(["tipo_caso", "condicion"]).size().unstack(fill_value=0)
pivot_cf.plot(kind="bar", ax=axes[3],
              color=[INS_VERDE, INS_ROJO, "#999999"][:len(pivot_cf.columns)], edgecolor="white")
axes[3].set_title("Condición final por tipo de caso")
axes[3].legend(fontsize=9)
axes[3].tick_params(axis="x", rotation=15)

fig.suptitle("Análisis Clínico — Magdalena 2019-2024", fontsize=15, fontweight="bold", y=1.01)
plt.tight_layout()
plt.savefig(DIR_FIGURAS / "analisis_clinico.png", dpi=150, bbox_inches="tight")
plt.show()

# Fuente de notificación ───────
FUENTE_LABELS = {
    "1": "Notificación rutinaria", "2": "Búsqueda activa institucional",
    "3": "Vigilancia intensificada", "4": "Búsqueda activa comunitaria", "5": "Investigaciones"
}

conteo_fuente = df_mag["fuente"].astype(str).str.strip().value_counts().reset_index()
conteo_fuente.columns = ["fuente", "casos"]
conteo_fuente["nombre"] = conteo_fuente["fuente"].map(FUENTE_LABELS).fillna(conteo_fuente["fuente"])

fig_f, ax_f = plt.subplots(figsize=(8, 5))
ax_f.pie(conteo_fuente["casos"], labels=conteo_fuente["nombre"],
         autopct="%1.1f%%", colors=PALETA[:len(conteo_fuente)])
ax_f.set_title("Fuente de notificación — Magdalena 2019-2024")
plt.tight_layout()
plt.savefig(DIR_FIGURAS / "fuente_notificacion.png", dpi=150, bbox_inches="tight")
plt.show()
```
